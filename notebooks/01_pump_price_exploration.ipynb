# Pump Price Prediction - Exploration

This notebook explores the processed dataset used to predict whether US gasoline prices increase the following week.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path('..').resolve() if Path.cwd().name == 'notebooks' else Path.cwd()
df = pd.read_csv(ROOT / 'data' / 'processed' / 'pump_price_dataset.csv', parse_dates=['date'])
df.shape
```

```python
df.head()
```

```python
df[['pump_price', 'brent_price', 'maritime_stress', 'target_price_up_1w']].describe()
```

```python
df['target_price_up_1w'].value_counts(normalize=True).rename('share')
```

```python
ax = df.plot(x='date', y=['pump_price', 'brent_price'], figsize=(12, 5), secondary_y='brent_price')
plt.title('Pump price and Brent over time')
plt.tight_layout()
```

```python
corr_cols = ['pump_return_1w', 'brent_return_1w', 'brent_return_4w', 'maritime_stress', 'strait_of_hormuz_n_tanker_ratio12', 'target_return_1w']
df[corr_cols].corr().round(3)
```
